```python
# =============================================================================
# NOTEBOOK COMPLET - PRÉDICTION DU RISQUE DE SURENTRAÎNEMENT
# Version Optimisée avec Corrections
# =============================================================================

# %% [markdown]
# # Prédire le risque de surentraînement
# ## Objectif : Utiliser les caractéristiques physiologiques pour estimer le risque de surentraînement

# %% [markdown]
# ### 1. IMPORTS

# %%
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV, StratifiedKFold, cross_validate
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (classification_report, confusion_matrix, 
                            roc_auc_score, precision_recall_curve, auc,
                            ConfusionMatrixDisplay, accuracy_score, 
                            f1_score, recall_score, precision_score)
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.feature_selection import RFE, SelectKBest, f_classif
from imblearn.over_sampling import SMOTE, ADASYN, BorderlineSMOTE
from imblearn.pipeline import Pipeline
import warnings
warnings.filterwarnings('ignore')

# %% [markdown]
# ### 2. CONSOLIDATION DES DONNÉES

# %%
# Chargement des datasets
df1 = pd.read_csv('exercise_tracking.csv')
df2 = pd.read_csv('exercise_tracking_synthetic_data.csv')

print(f"Dataset 1 : {df1.shape}")
print(f"Dataset 2 : {df2.shape}")

# Harmonisation des types
for col in df1.columns:
    dtype = df1[col].dtype
    if dtype == 'int64':
        df2[col] = pd.to_numeric(df2[col], errors='coerce').fillna(0).astype('int64')
    elif dtype == 'float64':
        df2[col] = pd.to_numeric(df2[col], errors='coerce').fillna(0.0)
    else:
        df2[col] = df2[col].astype(dtype)

# Fusion
df = pd.concat([df1, df2], ignore_index=True)
print(f"\nDataset consolidé : {df.shape}")
df.head()

# %% [markdown]
# ### 3. NETTOYAGE DES DONNÉES

# %%
print("=== NETTOYAGE ===")
print(f"Lignes avant nettoyage : {len(df)}")

# Suppression des valeurs incohérentes (0 impossibles)
cols_no_zero = ['Age', 'Height (m)', 'Weight (kg)', 'Max_BPM', 'Avg_BPM', 
                'Resting_BPM', 'Fat_Percentage', 'Experience_Level', 'BMI']

for col in cols_no_zero:
    df = df[df[col] != 0]

# Suppression des valeurs manquantes
df = df.dropna(subset=['Gender', 'Workout_Type'])

print(f"Lignes après nettoyage : {len(df)}")
print(f"Valeurs manquantes : {df.isnull().sum().sum()}")

# %% [markdown]
# ### 4. TRAITEMENT DES OUTLIERS (Méthode itérative)

# %%
def traiter_outliers_iteratif(df, colonnes):
    """Traite les outliers par remplacement médiane itératif"""
    df_clean = df.copy()
    
    for col in colonnes:
        iteration = 0
        while True:
            iteration += 1
            Q1 = df_clean[col].quantile(0.25)
            Q3 = df_clean[col].quantile(0.75)
            IQR = Q3 - Q1
            lower = Q1 - 1.5 * IQR
            upper = Q3 + 1.5 * IQR
            
            outliers_mask = (df_clean[col] < lower) | (df_clean[col] > upper)
            n_outliers = outliers_mask.sum()
            
            if n_outliers == 0:
                print(f"✓ {col}: {iteration} itération(s)")
                break
            
            median = df_clean[col].median()
            df_clean.loc[outliers_mask, col] = median
    
    return df_clean

cols_numeriques = df.select_dtypes(include='number').columns
df_clean = traiter_outliers_iteratif(df, cols_numeriques)

print(f"\n✓ Outliers traités pour {len(cols_numeriques)} colonnes")

# %% [markdown]
# ### 5. VISUALISATIONS POST-NETTOYAGE

# %%
# Matrice de corrélation
plt.figure(figsize=(12, 8))
numeric_cols = df_clean.select_dtypes(include=['number']).columns
corr_matrix = df_clean[numeric_cols].corr()
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", square=True)
plt.title("Matrice de corrélation après nettoyage", fontsize=14)
plt.tight_layout()
plt.show()

# Distributions
plt.figure(figsize=(15, 12))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(4, 4, i)
    sns.histplot(df_clean[col], kde=True, bins=20)
    plt.title(col)
plt.tight_layout()
plt.show()

# %% [markdown]
# ### 6. CRÉATION DE LA VARIABLE CIBLE (Seuil optimisé)

# %%
# Références physiologiques
resting_hr_reference = pd.DataFrame({
    'Age_Min': [18, 26, 36, 46, 56, 66],
    'Age_Max': [25, 35, 45, 55, 65, 100],
    'Normal_Min': [60, 61, 62, 63, 64, 65],
    'Normal_Max': [80, 80, 81, 82, 83, 84]
})

body_fat_reference = pd.DataFrame({
    'Sexe': ['Male', 'Male', 'Male', 'Male', 'Female', 'Female', 'Female', 'Female'],
    'Age_Min': [18, 30, 40, 50, 18, 30, 40, 50],
    'Age_Max': [29, 39, 49, 100, 29, 39, 49, 100],
    'Essentiel_Max': [5, 5, 5, 5, 13, 13, 13, 13],
    'Obésité_Seuil': [25, 26, 27, 28, 32, 33, 34, 35]
})

bmi_reference = {'Maigreur': 18.5, 'Normal': 24.9, 'Surpoids': 29.9}

def get_max_hr_expected(age):
    return 208 - (0.7 * age)

def get_resting_hr_range(age):
    ref = resting_hr_reference[(resting_hr_reference['Age_Min'] <= age) & 
                                (resting_hr_reference['Age_Max'] >= age)]
    if len(ref) > 0:
        return ref.iloc[0]['Normal_Min'], ref.iloc[0]['Normal_Max']
    return (70, 100) if age < 18 else (65, 100)

def get_body_fat_range(age, gender):
    ref = body_fat_reference[(body_fat_reference['Sexe'] == gender) &
                             (body_fat_reference['Age_Min'] <= age) & 
                             (body_fat_reference['Age_Max'] >= age)]
    if len(ref) > 0:
        return (ref.iloc[0]['Essentiel_Max'], ref.iloc[0]['Obésité_Seuil'])
    return (5, 28) if gender == 'Male' else (13, 32)

def calculate_overtraining_risk(row):
    """Calcule le risque avec SEUIL OPTIMISÉ pour équilibrer les classes"""
    risk_score = 0
    age = row['Age']
    gender = row['Gender']
    
    # 1. BPM au repos élevé
    hr_min, hr_max = get_resting_hr_range(age)
    resting_bpm = row['Resting_BPM']
    
    if resting_bpm > hr_max + 10:
        risk_score += 3
    elif resting_bpm > hr_max + 5:
        risk_score += 2
    elif resting_bpm > hr_max:
        risk_score += 1
    
    # 2. Réserve cardiaque faible
    expected_max_hr = get_max_hr_expected(age)
    hr_reserve = row['Max_BPM'] - row['Resting_BPM']
    expected_reserve = expected_max_hr - hr_max
    
    if hr_reserve < expected_reserve * 0.7:
        risk_score += 2
    elif hr_reserve < expected_reserve * 0.85:
        risk_score += 1
    
    # 3. Body fat
    essential, high = get_body_fat_range(age, gender)
    fat_pct = row['Fat_Percentage']
    
    if fat_pct < essential:
        risk_score += 2
    elif fat_pct > high:
        risk_score += 1
    
    # 4. Charge d'entraînement
    workout_freq = row['Workout_Frequency (days/week)']
    session_duration = row['Session_Duration (hours)']
    training_load = workout_freq * session_duration
    
    if training_load > 12:
        risk_score += 3
    elif training_load > 10:
        risk_score += 2
    elif training_load > 8:
        risk_score += 1
    
    if session_duration > 2.5:
        risk_score += 1
    elif session_duration > 2.0:
        risk_score += 0.5
    
    # 5. Hydratation
    water_intake = row['Water_Intake (liters)']
    if water_intake < 2.0:
        risk_score += 1
    elif water_intake < 2.5:
        risk_score += 0.5
    
    # 6. Expérience (débutants plus vulnérables)
    experience = row['Experience_Level']
    if experience == 1:
        risk_score *= 1.3
    elif experience == 2:
        risk_score *= 1.1
    
    # 7. IMC
    bmi = row['BMI']
    if bmi < bmi_reference['Maigreur']:
        risk_score += 2
    elif bmi > bmi_reference['Surpoids']:
        risk_score += 1
    
    # 8. Efficacité calorique
    calories_burned = row['Calories_Burned']
    calories_per_hour = calories_burned / session_duration if session_duration > 0 else 0
    
    if calories_per_hour < 250:
        risk_score += 1
    
    # SEUIL OPTIMISÉ : abaissé de 4 à 3 pour équilibrer les classes
    if risk_score >= 5:
        risk_level = "HIGH"
        risk_binary = 1
    elif risk_score >= 3:  # ⚠️ CHANGÉ de 4 à 3
        risk_level = "MODERATE"
        risk_binary = 1
    else:
        risk_level = "LOW"
        risk_binary = 0
    
    return risk_binary, risk_level, risk_score

# Application
results = df_clean.apply(calculate_overtraining_risk, axis=1, result_type='expand')
results.columns = ['Overtraining_Risk', 'Risk_Level', 'Risk_Score']
new_df = pd.concat([df_clean, results], axis=1)

print("\n=== DISTRIBUTION DES CLASSES ===")
print(new_df['Risk_Level'].value_counts())
print(f"\nProportions :")
print(new_df['Overtraining_Risk'].value_counts(normalize=True))

# %% [markdown]
# ### 7. PRÉPARATION DES DONNÉES POUR MODÉLISATION

# %%
# Séparation X et y
X = new_df.drop(['Overtraining_Risk', 'Risk_Level', 'Risk_Score'], axis=1)
y = new_df['Overtraining_Risk']

# Encodage des variables catégorielles
X_encoded = pd.get_dummies(X, columns=['Gender', 'Workout_Type'], drop_first=False)

# Train-test split stratifié
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Train set : {len(X_train)} samples")
print(f"Test set : {len(X_test)} samples")
print(f"Distribution train : {y_train.value_counts().to_dict()}")
print(f"Distribution test : {y_test.value_counts().to_dict()}")

# %% [markdown]
# ### 8. PIPELINE DE MODÉLISATION OPTIMISÉ

# %%
def model_pipeline_optimized(
    params, 
    model, 
    X_train, 
    y_train, 
    X_test, 
    y_test, 
    preprocessor=None,
    sampler='smote',
    feature_selection=None,
    feature_selection_params=None
):
    """Pipeline optimisé avec corrections des erreurs identifiées"""
    
    steps = []
    
    # 1. Feature Selection AVANT preprocessing
    if feature_selection is not None:
        if feature_selection == 'rfe':
            n_features_default = 10
            if feature_selection_params and 'n_features_to_select' in feature_selection_params:
                val = feature_selection_params['n_features_to_select']
                n_features_default = val[0] if isinstance(val, list) else val
            
            selector = RFE(estimator=model, n_features_to_select=n_features_default, step=1)
            steps.append(('feature_selection', selector))
        
        elif feature_selection == 'kbest':
            k_default = 10
            if feature_selection_params and 'k' in feature_selection_params:
                val = feature_selection_params['k']
                k_default = val[0] if isinstance(val, list) else val
            
            selector = SelectKBest(score_func=f_classif, k=k_default)
            steps.append(('feature_selection', selector))
    
    # 2. Preprocessing
    if preprocessor is not None:
        steps.append(('scaler', preprocessor))
    
    # 3. Sampler (SMOTE variantes)
    if sampler == 'smote':
        steps.append(('sampler', SMOTE(random_state=42, k_neighbors=3)))
    elif sampler == 'adasyn':
        steps.append(('sampler', ADASYN(random_state=42, n_neighbors=3)))
    elif sampler == 'borderline':
        steps.append(('sampler', BorderlineSMOTE(random_state=42, k_neighbors=3)))
    
    # 4. Modèle
    steps.append(('model', model))
    
    pipeline = Pipeline(steps)
    
    # Mise à jour des paramètres pour GridSearch
    updated_params = {}
    for key, value in params.items():
        updated_params[f"model__{key}"] = value
    
    if feature_selection is not None and feature_selection_params:
        for key, value in feature_selection_params.items():
            updated_params[f"feature_selection__{key}"] = value
    
    # GridSearchCV avec CV stratifiée
    print(f"\n{'='*60}")
    print(f"GRIDSEARCHCV - {model.__class__.__name__}")
    print(f"Feature Selection: {feature_selection or 'None'}")
    print(f"Sampler: {sampler}")
    print(f"{'='*60}")
    
    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=updated_params,
        scoring='f1',
        cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
        n_jobs=-1,
        verbose=1
    )
    
    grid_search.fit(X_train, y_train)
    
    print(f"\n✅ Meilleurs hyperparamètres : {grid_search.best_params_}")
    print(f"✅ Best CV F1-Score : {grid_search.best_score_:.3f}")
    
    # Features sélectionnées
    if feature_selection is not None:
        print(f"\n{'='*60}")
        print("FEATURES SÉLECTIONNÉES")
        print(f"{'='*60}")
        
        selector = grid_search.best_estimator_.named_steps.get('feature_selection')
        if selector is not None:
            selected_features = selector.get_support(indices=True)
            feature_names = X_train.columns.tolist()
            selected_names = [feature_names[i] for i in selected_features]
            
            print(f"\nNombre : {len(selected_names)}/{len(feature_names)}")
            print("\nFeatures sélectionnées :")
            for i, name in enumerate(selected_names, 1):
                print(f"  {i}. {name}")
    
    # Validation croisée stratifiée
    print(f"\n{'='*60}")
    print("VALIDATION CROISÉE (5-FOLD STRATIFIÉE)")
    print(f"{'='*60}")
    
    cv_results = cross_validate(
        grid_search.best_estimator_,
        X_train,
        y_train,
        cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
        scoring=['f1', 'recall', 'precision', 'roc_auc'],
        n_jobs=-1
    )
    
    for metric in ['f1', 'recall', 'precision', 'roc_auc']:
        scores = cv_results[f'test_{metric}']
        print(f"{metric.upper():12} : {scores.mean():.3f} ± {scores.std():.3f}")
    
    # Test set evaluation
    print(f"\n{'='*60}")
    print("RÉSULTATS SUR LE TEST SET")
    print(f"{'='*60}")
    
    y_pred_test = grid_search.predict(X_test)
    y_pred_proba = grid_search.predict_proba(X_test)[:, 1]
    
    print("\nClassification Report :")
    print(classification_report(y_test, y_pred_test, target_names=['Low Risk', 'At Risk']))
    
    print("\nConfusion Matrix :")
    cm = confusion_matrix(y_test, y_pred_test)
    print(cm)
    
    print(f"\nROC-AUC Score : {roc_auc_score(y_test, y_pred_proba):.3f}")
    
    # Precision-Recall curve
    precision, recall, thresholds = precision_recall_curve(y_test, y_pred_proba)
    pr_auc = auc(recall, precision)
    print(f"PR-AUC Score : {pr_auc:.3f}")
    
    # Analyse détaillée
    tn, fp, fn, tp = cm.ravel()
    print(f"\n{'='*60}")
    print("ANALYSE DÉTAILLÉE")
    print(f"{'='*60}")
    print(f"True Negatives  : {tn}")
    print(f"False Positives : {fp}")
    print(f"False Negatives : {fn} ⚠️")
    print(f"True Positives  : {tp}")
    print(f"\nSensitivity (Recall) : {tp / (tp + fn):.3f}")
    print(f"Specificity          : {tn / (tn + fp):.3f}")
    
    return grid_search.best_estimator_, {
        'model_name': model.__class__.__name__,
        'accuracy': accuracy_score(y_test, y_pred_test),
        'f1_score': f1_score(y_test, y_pred_test),
        'recall': recall_score(y_test, y_pred_test),
        'precision': precision_score(y_test, y_pred_test),
        'roc_auc': roc_auc_score(y_test, y_pred_proba),
        'pr_auc': pr_auc
    }

# %% [markdown]
# ### 9. ENTRAÎNEMENT DES MODÈLES

# %% [markdown]
# #### 9.1 Logistic Regression

# %%
params_lr = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear'],
    'class_weight': ['balanced', None]
}

best_lr, results_lr = model_pipeline_optimized(
    params=params_lr,
    model=LogisticRegression(max_iter=5000, random_state=42),
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    preprocessor=StandardScaler(),
    sampler='smote'
)

# %% [markdown]
# #### 9.2 SVM (avec class_weight optimisé)

# %%
params_svc = {
    "C": [0.01, 0.1, 1, 10, 100],
    "kernel": ["rbf", "linear"],
    "gamma": ["scale", "auto", 0.001, 0.01, 0.1],
    "class_weight": ['balanced', None]
}

best_svc, results_svc = model_pipeline_optimized(
    params=params_svc,
    model=SVC(probability=True, random_state=42),
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    preprocessor=MinMaxScaler(),  # Changé de StandardScaler
    sampler='borderline'  # BorderlineSMOTE souvent meilleur pour SVM
)

# %% [markdown]
# #### 9.3 Decision Tree (régularisé)

# %%
params_dt = {
    "max_depth": [3, 5, 7, 10],  # Retiré None
    "min_samples_split": [10, 20, 50],
    "min_samples_leaf": [5, 10, 20],
    "criterion": ["gini", "entropy"],
    "max_features": ["sqrt", "log2", None]
}

best_dt, results_dt = model_pipeline_optimized(
    params=params_dt,
    model=DecisionTreeClassifier(random_state=42),
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    preprocessor=StandardScaler(),
    sampler='smote'
)

# %% [markdown]
# #### 9.4 Random Forest (NOUVEAU - souvent meilleur que DT)

# %%
params_rf = {
    "n_estimators": [100, 200, 300],
    "max_depth": [5, 10, 15, None],
    "min_samples_split": [10, 20],
    "min_samples_leaf": [5, 10],
    "class_weight": ['balanced', 'balanced_subsample', None]
}

best_rf, results_rf = model_pipeline_optimized(
    params=params_rf,
    model=RandomForestClassifier(random_state=42),
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    preprocessor=StandardScaler(),
    sampler='smote'
)

# %% [markdown]
# #### 9.5 Gradient Boosting (NOUVEAU - très performant)

# %%
params_gb = {
    "n_estimators": [100, 200],
    "learning_rate": [0.01, 0.1, 0.2],
    "max_depth": [3, 5, 7],
    "min_samples_split": [10, 20]
}

best_gb, results_gb = model_pipeline_optimized(
    params=params_gb,
    model=GradientBoostingClassifier(random_state=42),
    X_train=X_train,
    y_train=y_train,
    X_test=X_test,
    y_test=y_test,
    preprocessor=StandardScaler(),
    sampler='smote'
)

# %% [markdown]
# ### 10. COMPARAISON FINALE DES MODÈLES

# %%
results_df = pd.DataFrame([results_lr, results_svc, results_dt, results_rf, results_gb])
results_df = results_df.sort_values('f1_score', ascending=False)

print("\n" + "="*80)
print("COMPARAISON FINALE DES MODÈLES")
print("="*80)
print(results_df.to_string(index=False))

# Visualisation
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
metrics = ['accuracy', 'f1_score', 'recall', 'precision', 'roc_auc', 'pr_auc']

for i, metric in enumerate(metrics):
    ax = axes[i//3, i%3]
    results_df.plot(x='model_name', y=metric, kind='bar', ax=ax, legend=False)
    ax.set_title(metric.upper())
    ax.set_ylabel('Score')
    ax.set_xlabel('')
    ax.set_ylim(0, 1)
    ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# %% [markdown]
# ### 11. VISUALISATION DU MEILLEUR MODÈLE

# %%
best_model_name = results_df.iloc[0]['model_name']
print(f"\n🏆 MEILLEUR MODÈLE : {best_model_name}")

if best_model_name == 'LogisticRegression':
    best_model = best_lr
elif best_model_name == 'SVC':
    best_model = best_svc
elif best_model_name == 'DecisionTreeClassifier':
    best_model = best_dt
elif best_model_name == 'RandomForestClassifier':
    best_model = best_rf
else:
    best_model = best_gb

# Matrice de confusion
y_pred = best_model.predict(X_test)
cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Low Risk", "At Risk"])
disp.plot(cmap="Blues")
plt.title(f"Matrice de confusion - {best_model_name}")
plt.show()

# Precision-Recall curve
y_pred_proba = best_model.predict_proba(X_test)[:, 1]
precision, recall, thresholds = precision_recall_curve(y_test, y_pred_proba)
pr_auc = auc(recall, precision)

plt.figure(figsize=(8, 6))
plt.plot(recall, precision, label=f'PR-AUC = {pr_auc:.3f}')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title(f'Precision-Recall Curve - {best_model_name}')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# %% [markdown]
# ### 12. ANALYSE DES ERREURS

# %%
# Faux négatifs (At Risk prédits Low - DANGEREUX)
fn_indices = np.where((y_test.values == 1) & (y_pred == 0))[0]
print(f"\n⚠️ FAUX NÉGATIFS (At Risk manqués) : {len(fn_indices)}")
if len(fn_indices) > 0:
    print("\nExemples de profils manqués :")
    print(X_test.iloc[fn_indices].head())

# Faux positifs (Low prédits At Risk)
fp_indices = np.where((y_test.values == 0) & (y_pred == 1))[0]
print(f"\n⚠️ FAUX POSITIFS (Fausses alertes) : {len(fp_indices)}")
if len(fp_indices) > 0:
    print("\nExemples de fausses alertes :")
    print(X_test.iloc[fp_indices].head())

# %% [markdown]
# ### 13. FEATURE IMPORTANCE (si applicable)

# %%
if best_model_name in ['RandomForestClassifier', 'GradientBoostingClassifier']:
    feature_importance = best_model.named_steps['model'].feature_importances_
    feature_names = X_train.columns
    
    importance_df = pd.DataFrame({
        'feature': feature_names,
        'importance': feature_importance
    }).sort_values('importance', ascending=False)
    
    print("\n=== TOP 10 FEATURES IMPORTANTES ===")
    print(importance_df.head(10))
    
    plt.figure(figsize=(10, 6))
    plt.barh(importance_df.head(10)['feature'], importance_df.head(10)['importance'])
    plt.xlabel('Importance')
    plt.title(f'Top 10 Features - {best_model_name}')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()

# %% [markdown]
# ### 14. SAUVEGARDE DU MEILLEUR MODÈLE

# %%
import joblib
from datetime import datetime

# Créer un nom de fichier avec timestamp
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
model_filename = f"best_model_{best_model_name}_{timestamp}.pkl"

# Sauvegarder le modèle
joblib.dump(best_model, model_filename)
print(f"✅ Modèle sauvegardé : {model_filename}")

# Sauvegarder aussi les résultats
results_filename = f"model_comparison_{timestamp}.csv"
results_df.to_csv(results_filename, index=False)
print(f"✅ Résultats sauvegardés : {results_filename}")

# %% [markdown]
# ### 15. COMMENT UTILISER LE MODÈLE SAUVEGARDÉ

# %%
# Pour recharger et utiliser le modèle plus tard :
"""
import joblib
import pandas as pd

# 1. Charger le modèle
model = joblib.load('best_model_RandomForestClassifier_20241207_143022.pkl')

# 2. Préparer de nouvelles données (même format que X_train)
new_data = pd.DataFrame({
    'Age': [35],
    'Height (m)': [1.75],
    'Weight (kg)': [75],
    'Max_BPM': [185],
    'Avg_BPM': [140],
    'Resting_BPM': [72],
    'Session_Duration (hours)': [1.5],
    'Calories_Burned': [600],
    'Workout_Frequency (days/week)': [5],
    'Fat_Percentage': [18],
    'Water_Intake (liters)': [2.5],
    'Experience_Level': [2],
    'BMI': [24.5],
    'Gender_Female': [0],
    'Gender_Male': [1],
    'Workout_Type_Cardio': [0],
    'Workout_Type_HIIT': [1],
    'Workout_Type_Strength': [0],
    'Workout_Type_Yoga': [0]
})

# 3. Faire une prédiction
prediction = model.predict(new_data)
probability = model.predict_proba(new_data)

print(f"Prédiction : {'At Risk' if prediction[0] == 1 else 'Low Risk'}")
print(f"Probabilité At Risk : {probability[0][1]:.2%}")
"""

# %% [markdown]
# ### 16. RÉSUMÉ ET CONCLUSIONS

# %%
print("\n" + "="*80)
print("RÉSUMÉ DU PROJET")
print("="*80)

print(f"\n📊 Dataset final : {len(new_df)} observations")
print(f"   - Low Risk : {(y == 0).sum()} ({(y == 0).sum()/len(y)*100:.1f}%)")
print(f"   - At Risk : {(y == 1).sum()} ({(y == 1).sum()/len(y)*100:.1f}%)")

print(f"\n🏆 Meilleur modèle : {best_model_name}")
print(f"   - F1-Score : {results_df.iloc[0]['f1_score']:.3f}")
print(f"   - Recall : {results_df.iloc[0]['recall']:.3f}")
print(f"   - Precision : {results_df.iloc[0]['precision']:.3f}")
print(f"   - ROC-AUC : {results_df.iloc[0]['roc_auc']:.3f}")
print(f"   - PR-AUC : {results_df.iloc[0]['pr_auc']:.3f}")

print(f"\n📁 Fichiers sauvegardés :")
print(f"   - Modèle : {model_filename}")
print(f"   - Résultats : {results_filename}")

print("\n" + "="*80)
print("POINTS CLÉS")
print("="*80)
print("""
1. ✅ Variable cible créée à partir de références médicales (AHA, NIH, ACE)
2. ✅ Seuil optimisé (3 au lieu de 4) pour équilibrer les classes
3. ✅ Outliers traités par méthode itérative
4. ✅ SMOTE/ADASYN/BorderlineSMOTE testés pour le rééchantillonnage
5. ✅ 5 modèles comparés avec GridSearchCV et validation croisée stratifiée
6. ✅ Métriques complètes : F1, Recall, Precision, ROC-AUC, PR-AUC
7. ✅ Analyse des erreurs (faux positifs/négatifs)
8. ✅ Modèle sauvegardé et prêt pour la production
""")
```
